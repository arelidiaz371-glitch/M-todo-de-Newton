1. Formulación del problema

El Ejemplo 1 del libro de Burden considera la ecuación:

$$f(x)=\cos x-x=0$$

El objetivo es encontrar una aproximación de la raíz de esta función.

Para aplicar el método de Newton, se necesita calcular la función y su primera derivada.

La función original es:

$$f(x)=\cos x-x$$

Su derivada es:

$$f’(x)=-\sin x-1$$

Fórmula del método de Newton

De acuerdo con la ecuación (2.7) de Burden, las aproximaciones se calculan mediante:

$$p_n=p_{n-1}-\frac{f(p_{n-1})}{f’(p_{n-1})}$$

Sustituyendo la función y su derivada:

$$p_n=p_{n-1}-\frac{\cos(p_{n-1})-p_{n-1}}{-\sin(p_{n-1})-1}$$

El proceso comienza con una aproximación inicial y continúa hasta que la diferencia absoluta entre dos aproximaciones consecutivas sea menor que la tolerancia establecida.

2. Descripción del algoritmo

El método de Newton se desarrolla siguiendo los pasos del Algoritmo 2.3 del libro de Burden:

1. Establecer la aproximación inicial (p_0), la tolerancia (TOL) y el número máximo de iteraciones (N_0).
2. Inicializar el contador de iteraciones en (i=1).
3. Calcular una nueva aproximación utilizando la fórmula de Newton.
4. Verificar si la diferencia absoluta entre la aproximación nueva y la anterior es menor que la tolerancia.
5. Si se cumple la condición, devolver la aproximación y finalizar el procedimiento.
6. Si no se cumple, incrementar el contador, actualizar la aproximación y repetir.
7. Si se alcanza el número máximo de iteraciones sin cumplir la tolerancia, informar que el método falló.

La condición de parada utilizada es:

$$|p-p_0|<TOL$$

```python
import math

def newton(p0, TOL, N0):
    """
    Metodo de Newton para aproximar una raiz.
    p0: aproximacion inicial
    TOL: tolerancia
    N0: numero maximo de iteraciones
    """

    i = 1

    print("METODO DE NEWTON")
    print("-" * 55)
    print(f"{'Iteracion':<12}{'Aproximacion':<25}{'Error':<18}")
    print("-" * 55)

    while i <= N0:
        # Evaluar la funcion y su derivada
        f = math.cos(p0) - p0
        df = -math.sin(p0) - 1

        # Comprobar que la derivada no sea cero
        if df == 0:
            print("Error: la derivada es cero.")
            return None

        # Calcular la nueva aproximacion
        p = p0 - f / df

        # Calcular el error absoluto
        error = abs(p - p0)

        # Mostrar los resultados de la iteracion
        print(f"{i:<12}{p:<25.10f}{error:<18.10f}")

        # Verificar el criterio de parada
        if error < TOL:
            print("-" * 55)
            print("El procedimiento fue exitoso.")
            print(f"Raiz aproximada: {p:.10f}")
            print(f"Iteraciones realizadas: {i}")
            print(f"Tolerancia: {TOL}")
            return p

        # Actualizar el contador y la aproximacion
        i = i + 1
        p0 = p

    print("-" * 55)
    print("El metodo fallo al alcanzar el maximo de iteraciones.")
    return None


# Datos iniciales del problema
p0 = 0.5
TOL = 1e-5
N0 = 100

# Ejecutar el metodo de Newton
resultado = newton(p0, TOL, N0)

import math

def newton(p0, TOL, N0):
    """
    Metodo de Newton para aproximar una raiz.
    p0: aproximacion inicial
    TOL: tolerancia
    N0: numero maximo de iteraciones
    """

    i = 1

    print("METODO DE NEWTON")
    print("-" * 55)
    print(f"{'Iteracion':<12}{'Aproximacion':<25}{'Error':<18}")
    print("-" * 55)

    while i <= N0:
        # Evaluar la funcion y su derivada
        f = math.cos(p0) - p0
        df = -math.sin(p0) - 1

        # Comprobar que la derivada no sea cero
        if df == 0:
            print("Error: la derivada es cero.")
            return None

        # Calcular la nueva aproximacion
        p = p0 - f / df

        # Calcular el error absoluto
        error = abs(p - p0)

        # Mostrar los resultados de la iteracion
        print(f"{i:<12}{p:<25.10f}{error:<18.10f}")

        # Verificar el criterio de parada
        if error < TOL:
            print("-" * 55)
            print("El procedimiento fue exitoso.")
            print(f"Raiz aproximada: {p:.10f}")
            print(f"Iteraciones realizadas: {i}")
            print(f"Tolerancia: {TOL}")
            return p

        # Actualizar el contador y la aproximacion
        i = i + 1
        p0 = p

    print("-" * 55)
    print("El metodo fallo al alcanzar el maximo de iteraciones.")
    return None


# Datos iniciales del problema
p0 = 0.5
TOL = 1e-5
N0 = 100

# Ejecutar el metodo de Newton
resultado = newton(p0, TOL, N0)
```
